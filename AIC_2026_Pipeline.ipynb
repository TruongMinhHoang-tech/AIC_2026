```python
import cv2
import numpy as np
from PIL import Image
from paddleocr import PaddleOCR
from vietocr.tool.predictor import Predictor
from vietocr.tool.config import Cfg


paddle_detector = PaddleOCR(use_angle_cls=True, lang='vi', rec=True, show_log=False, use_gpu=False, det_limit_side_len=1536, det_db_unclip_ratio=2.0)
config = Cfg.load_config_from_name('vgg_transformer')
config['device'] = 'cpu'
viet_recognizer = Predictor(config)


def extract_text_from_image(img_path):
    img_cv = cv2.imread(img_path)
    if img_cv is None:
        return ""

    boxes = paddle_detector.ocr(img_path, cls=True)
    text_lines = []

    if boxes and boxes[0] is not None:
        for item in boxes[0]:
            box, paddle_text = item[0], item[1][0]


            if not any(char.isalpha() for char in paddle_text):
                text_lines.append(paddle_text)
            else:
                x_coords, y_coords = [int(p[0]) for p in box], [int(p[1]) for p in box]
                cropped_img = img_cv[min(y_coords):max(y_coords), min(x_coords):max(x_coords)]

                if cropped_img.size > 0:

                    gray = cv2.cvtColor(cropped_img, cv2.COLOR_BGR2GRAY)
                    contrast_img = cv2.createCLAHE(clipLimit=1.0, tileGridSize=(8, 8)).apply(gray)
                    final_cv_img = cv2.cvtColor(contrast_img, cv2.COLOR_GRAY2RGB)

                    viet_text = viet_recognizer.predict(Image.fromarray(final_cv_img))
                    text_lines.append(viet_text)

    return " | ".join(text_lines)

```

```python
import os
import glob
import csv


folder_path = 'L21_V001'
output_csv = f'ocr_results_{folder_path}.csv'
video_id = os.path.basename(os.path.normpath(folder_path))


image_list = glob.glob(os.path.join(folder_path, '*.[jp][pn]g'))
print(f"Bắt đầu quét {len(image_list)} ảnh từ Video {video_id}...\n")


processed_frames = set()
if os.path.isfile(output_csv):
    with open(output_csv, 'r', encoding='utf-8-sig') as f:
        reader = csv.reader(f)
        next(reader, None)
        for row in reader:
            if len(row) >= 2:
                processed_frames.add(row[1])


with open(output_csv, 'a', newline='', encoding='utf-8-sig') as csvfile:
    writer = csv.writer(csvfile)


    if os.stat(output_csv).st_size == 0 if os.path.isfile(output_csv) else True:
        writer.writerow(['video_id', 'frame_id', 'text_doc_duoc'])

    for img_path in image_list:
        frame_id = os.path.splitext(os.path.basename(img_path))[0]

        if frame_id in processed_frames:
            continue

        print(f" Đang quét: Frame [{frame_id}]")


        text_result = extract_text_from_image(img_path)


        writer.writerow([video_id, frame_id, text_result])
        csvfile.flush()

print(f"\n HOÀN THÀNH! Đã lưu kết quả tại: {output_csv}")
```

```python
import csv
import re


def load_ocr_data(csv_filepath):
    ocr_data = []
    with open(csv_filepath, 'r', encoding='utf-8-sig') as f:
        for row in csv.DictReader(f):
            ocr_data.append({
                'video_id': row['video_id'],
                'frame_id': row['frame_id'],
                'text': row['text_doc_duoc']
            })
    return ocr_data


def ocr_search(query: str, ocr_data: list, min_score=60):
    results = []
    query_clean = query.lower().strip()


    query_words = set(query_clean.split())

    for item in ocr_data:
        text = item.get('text', '').lower()
        if not text:
            continue


        if query_clean in text:
            results.append((item['video_id'], item['frame_id'], 100.0, item['text']))
            continue


        if query_words:

            text_words = set(text.replace('|', ' ').split())


            matched_words = query_words.intersection(text_words)
            score = (len(matched_words) / len(query_words)) * 100


            if score >= min_score:
                results.append((item['video_id'], item['frame_id'], score, item['text']))


    results.sort(key=lambda x: x[2], reverse=True)
    return results


file_csv = 'ocr_results_L21_V001.csv'


tu_khoa = "Thái nguyên: khẩn trương khắc phục hậu quả sau mưa to"

try:
    du_lieu_ocr = load_ocr_data(file_csv)
    print(f"Đang tìm từ khóa: '{tu_khoa}'...")


    ket_qua = ocr_search(tu_khoa, du_lieu_ocr, min_score=50)

    if not ket_qua:
        print("Không tìm thấy khung hình nào phù hợp.")
    else:
        print(f"Tìm thấy {len(ket_qua)} kết quả phù hợp (Đã sắp xếp theo độ chính xác):")
        for idx, (vid, fid, score, text) in enumerate(ket_qua):

            print(f"\n[{idx + 1}] Độ khớp {score:.0f}% | Video: {vid} - Frame: {fid}")
            print(f"    >> Văn bản gốc: {text[:150]}...")

except FileNotFoundError:
    print(f"[LỖI]: Không tìm thấy file '{file_csv}'!")
```
